```python
import json
import numpy as np

with open("./testdistance_estimates.json", encoding="utf-8") as f:
    data = json.load(f)

THRESH = 5.0
WINDOW = 3  # 前後の中央値を見る範囲

for scene_id, frames in data.items():
    keys = sorted(frames.keys())
    values = [frames[k] for k in keys]

    smoothed = []
    for i in range(len(values)):
        # 周囲のフレームインデックス
        start = max(0, i - WINDOW)
        end = min(len(values), i + WINDOW + 1)

        neighbors = values[start:i] + values[i+1:end]
        if len(neighbors) < 2:
            smoothed.append(values[i])
            continue

        median_val = np.median(neighbors)
        if abs(values[i] - median_val) > THRESH:
            smoothed.append(float(median_val))  # スパイクを中央値で補正
        else:
            smoothed.append(values[i])  # 問題なければそのまま

    # 上書き保存
    for k, v in zip(keys, smoothed):
        frames[k] = v

# 書き出し
with open("./testdistance_estimates_smoothed.json", "w", encoding="utf-8") as f:
    json.dump(data, f, indent=2, ensure_ascii=False)

print("✅ 中央値ベース + 局所スパイク除去処理を実行し、保存しました。")
```

```python
import os
import json
import matplotlib.pyplot as plt

# 読み込み
with open("./testdistance_estimates_smoothed.json", encoding="utf-8") as f:
    data = json.load(f)

# 出力ディレクトリ
output_dir = "./scene2"
os.makedirs(output_dir, exist_ok=True)

for scene_id, frame_distances in data.items():
    # フレーム番号順に並べ替え
    frames = sorted(frame_distances.keys())
    distances = [frame_distances[frame] for frame in frames]

    # プロット
    plt.figure()
    plt.plot(range(1, len(distances)+1), distances, marker='o')
    plt.title(f"Scene {scene_id} - Distance Estimate")
    plt.xlabel("Frame Index")
    plt.ylabel("Distance (m)")
    plt.grid(True)

    # 保存
    save_path = os.path.join(output_dir, f"{scene_id}.png")
    plt.savefig(save_path)
    plt.close()

print("✅ 各シーンのグラフを test/outputs/distance_graphs/ に保存しました。")
```
